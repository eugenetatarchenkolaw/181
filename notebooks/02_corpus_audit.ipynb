# Анализ корпуса второго чекпойнта

Это заготовка, а не выполненный анализ. Данные пока не собраны; ячейки не содержат результатов. Основной сбор выполняется переиспользуемыми скриптами. Здесь будут расчёты и объяснения для отчёта.

См. [задание](../docs/checkpoints/checkpoint-02.md), [схему](../data/checkpoint02/SCHEMA.md) и [шаблон отчёта](../reports/checkpoint02/REPORT_TEMPLATE.md).

## 1. Окружение и версия данных

Указать коммит кода, версию набора, Python и зависимости. Загрузить CSV/JSONL через стандартные средства; проверить, что формы наполнены. Остановить анализ с понятным сообщением, если данных ещё нет.

## 2. Основные статьи и библиография

Показать причины выбора минимум пяти работ. Сопоставить число записей в PDF с извлечённым списком. Посчитать статусы по каждому родителю. Показать ручные исправления отдельно.

## 3. Доступность источников

Посчитать автоматические и ручные загрузки, недоступные файлы и ошибки. Разделить число библиографических записей и уникальных PDF. Не удалять неуспешные записи из знаменателя.

## 4. Текст и связи

Проверить читаемость страниц, порядок чтения, общие источники и версии. Показать несколько конкретных ошибок парсинга и сопоставления. Проверить цепочку от родителя до доказательства.

## 5. Вопросы и доказательства

Посчитать draft/approved/rejected, типы вопросов, уникальные факты и фрагменты. Разобрать один принятый и один отклонённый пример. Проверить, что ответы следуют из цитат и что разметка не входит в будущий индекс.

## 6. Фиксация и выводы

После проверки заполнить манифест версии и отчёт. Указать фактические ограничения и задачи для BM25. Не подставлять ожидаемые метрики поиска вместо ещё не выполненного эксперимента.
